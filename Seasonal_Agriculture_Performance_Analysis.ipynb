## 10. Statistical Analysis

Statistical tests are applied to determine whether the observed differences in agricultural performance across seasons are statistically significant.


### 10.1 ANOVA: Yield Across Seasons

A one-way ANOVA test is used to determine whether the mean crop yield differs significantly across Kharif, Rabi, and Zaid seasons.

```python
from scipy.stats import f_oneway

groups = [group["Yield_Tonnes_Ha"].dropna() for _, group in df.groupby("Season")]

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

### 10.2 ANOVA: Profit Across Seasons

A one-way ANOVA test is used to determine whether the mean profit differs significantly across Kharif, Rabi, and Zaid seasons.

```python
groups = [group["Profit_INR"].dropna() for _, group in df.groupby("Season")]

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

### 10.3 ANOVA: Water Usage Across Seasons

A one-way ANOVA test is used to determine whether the mean water usage differs significantly across Kharif, Rabi, and Zaid seasons.

```python
groups = [group["Water_Used_m3"].dropna() for _, group in df.groupby("Season")]

f_stat, p_value = f_oneway(*groups)

print("F-statistic:", f_stat)
print("p-value:", p_value)
```

## 12. Key Findings

The analysis reveals several important patterns and variations in agricultural performance across seasons, crops, and regions.

### 12.1 Seasonal Yield Performance

Kharif recorded the highest average yield at 5.64 tonnes per hectare, followed by Rabi at 5.08 tonnes per hectare and Zaid at 4.67 tonnes per hectare. However, the ANOVA test indicated that the differences in mean yield across seasons were not statistically significant at the 5% significance level.

### 12.2 Seasonal Profitability

Kharif showed the strongest average profit at ₹178,914.65, followed by Rabi at ₹87,689.47. Zaid recorded an average loss of ₹24,804.82. The ANOVA test showed that the differences in mean profit across seasons were statistically significant.

### 12.3 Seasonal Water Usage and Efficiency

Zaid recorded the highest average water usage at 6,419.89 m³ and the lowest water efficiency at 4.41 tonnes per 1,000 m³. Kharif showed the highest water efficiency at 5.89 tonnes per 1,000 m³. However, the ANOVA test indicated that differences in average water usage across seasons were not statistically significant at the 5% significance level.

### 12.4 Rainfall and Yield Relationship

Rainfall varied considerably across seasons, with Kharif receiving the highest average rainfall. However, the correlation between rainfall and yield was close to zero both overall and within individual seasons, indicating that rainfall alone did not strongly explain yield variation in this dataset.

### 12.5 Crop-wise Seasonal Performance

All eight crops recorded their highest average yield in Kharif, followed by Rabi and Zaid. This consistent pattern indicates that seasonal conditions are associated with clear differences in crop performance, although the magnitude of variation differs across crops.

### 12.6 Regional and Seasonal Variation

Seasonal performance was not uniform across all states. While Kharif generally showed strong performance, Punjab recorded its highest average yield in Rabi, and Karnataka recorded its highest average yield in Zaid. This indicates that regional conditions can influence seasonal agricultural performance.

### 12.7 Crop Profitability Variation

Profitability varied substantially across crops and seasons. Sugarcane and Chilli showed relatively strong profitability across seasons, while several other crops recorded negative average profits, particularly in Zaid. This indicates that higher production or yield does not necessarily result in higher profitability.

## 13. Recommendations

Based on the findings from the seasonal, crop-wise, resource, environmental, and regional analysis, the following recommendations can support better agricultural planning and resource management.

### 13.1 Season-specific Agricultural Planning

Agricultural planning should consider seasonal differences in profitability, resource usage, and crop performance. Kharif showed the strongest average profitability, while Zaid recorded an average loss, suggesting the need for more careful planning during the Zaid season.

### 13.2 Improve Water Management in Zaid

Zaid recorded the highest average water usage and the lowest water efficiency among the three seasons. Therefore, efficient irrigation practices, better water allocation, and regular monitoring of water usage should be considered during the Zaid season.

### 13.3 Select Crops Based on Profitability

Crop selection should consider profitability along with yield and production. The analysis shows that some crops can have good production but still generate negative average profit. Therefore, profitability should be considered when making seasonal crop planning decisions.

### 13.4 Adopt Region-specific Strategies

Agricultural strategies should be adapted to regional conditions because seasonal performance varies across states. For example, Punjab showed its highest average yield in Rabi, while Karnataka showed its highest average yield in Zaid. Region-specific planning can therefore support better seasonal outcomes.

### 13.5 Consider Multiple Factors in Decision-making

Rainfall and soil moisture showed weak linear relationships with yield in this dataset. Therefore, agricultural decisions should not rely on a single environmental factor. Crop type, season, water usage, regional conditions, and economic factors should be considered together.

## 14. Conclusion

The analysis provides a clear view of how agricultural performance varies across seasons, crops, and regions. The findings highlight differences in yield, profitability, water usage, and environmental conditions, while also showing that seasonal and regional factors should be considered together for effective agricultural planning.
